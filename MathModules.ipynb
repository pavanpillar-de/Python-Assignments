```python
#Reverse a three-digit number using arithmetic operators.

num = 456

hundreds = num // 100
tens = (num // 10) % 10
units = num % 10

rev_num = (units * 100) + (tens * 10) + hundreds

print(f"Reversed number: {rev_num}")

```

```python
#Find the last digit and first digit of a given number
n = 764

last_digit = n % 10

first_digit = n
while first_digit >= 10:
    first_digit //= 10

print(f"First digit: {first_digit}")
print(f"Last digit: {last_digit}")
```

```python
#Count how many vowels in a string belong to a predefined vowel list using in.
vowels = ['a', 'e', 'i', 'o', 'u']

text = input("Enter a string: ")
count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print(f"Number of vowels in '{text}': {count}")
```

```python
#Build a simple calculator using arithmetic and comparison operators


num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
  if num2 != 0:
    result = num1 / num2
  else:
    result = "Error: Division by zero"
print(f"result:{result}")
```

```python
#Create a grading system using comparison and logical operators.
marks = float(input("Enter marks: "))

if marks >= 90:
    grade = "A+"
elif marks >= 80 and marks < 90:
    grade = "A"
elif marks >= 70 and marks < 80:
    grade = "B"
elif marks >= 60 and marks < 70:
    grade = "C"
elif marks >= 50 and marks < 60:
    grade = "D"
else:
    grade = "F"
if 0 <= marks <= 100:
    print(f"Marks: {marks}")
    print(f"Grade: {grade}")
```
